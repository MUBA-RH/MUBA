```python
ROOT=pathlib.Path('/kaggle/working/MUBA')
if not ROOT.exists():
    subprocess.check_call(['git','clone','--depth','1','https://github.com/MUBA-RH/MUBA.git',str(ROOT)])
else:
    subprocess.check_call(['git','pull','--ff-only'],cwd=ROOT)
WORKER=ROOT/'daily-story-worker'
assert WORKER.exists(), 'daily-story-worker not found'
print('MUBA worker:', WORKER)
```
